```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Number of data qubits
n_data = 49

# Select one Z-type stabilizer
z_check = [0, 1, 7, 8]

# Ancilla is qubit 49
ancilla = 49

# 49 data qubits + 1 ancilla + 1 classical bit
qc = QuantumCircuit(n_data + 1, 1)

qc.x(0)  # Inject an X error on data qubit 0 

# Step 1: All qubits start in |0>

# Step 2: Measure the Z stabilizer using CNOT gates
for q in z_check:
    qc.cx(q, ancilla)

# Step 3: Measure the ancilla
qc.measure(ancilla, 0)

# Display circuit
print(qc.draw())

# Run simulation using the stabilizer method
simulator = AerSimulator(method="stabilizer")

# CHANGE: Run qc directly; no transpile() needed here
result = simulator.run(qc, shots=100).result()

counts = result.get_counts()
print("Measurement results:", counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

n_data = 49
ancilla = 49
x_check = [0, 1]

qc_x = QuantumCircuit(n_data + 1, 1)

# Prepare all data qubits in |+>
# This makes X0 X1 have eigenvalue +1.
for q in range(n_data):
    qc_x.h(q)

# Prepare ancilla in |+>
qc_x.h(ancilla)

# Ancilla controls the data qubits in the X-check
for q in x_check:
    qc_x.cx(ancilla, q)

# Change ancilla measurement from X basis to Z basis
qc_x.h(ancilla)

# Measure only the ancilla
qc_x.measure(ancilla, 0)

# Simulate using the stabilizer method
simulator = AerSimulator(method="stabilizer")
result = simulator.run(qc_x, shots=100).result()

print("X-check results:", result.get_counts())
```
