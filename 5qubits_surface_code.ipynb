```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator

# ======================================================
# STEP 1: Data qubits and the two stabilizer layouts
# ======================================================
n_data = 25
#   q0 q1 q2
#   q3 q4 q5
#   q6 q7 q8

LAYOUT_A = {
    "x": [[0, 1, 5, 6], [1,2], [2,3,7,8], [3,4], [6,7,11,12], [8,9,13,14], [10,11,15,16], [12,13,17,18], [16,17,21,22], [18,19,23,24], [20, 21], [22,23]],   # X-type stabilizers
    "z": [[0,5], [1,2,6,7], [3,4,8,9], [5,6,10,11], [7,8,12,13], [9,14], [10,15], [11,12,16,17], [13,14,18,19], [15,16,20,21], [17,18,22,23], [19,24]],   # Z-type stabilizers
}
LAYOUT_B = {
    "x": [[0, 1, 5, 6], [2,3,7,8], [4,9], [5,10], [6,7,11,12], [8,9,13,14], [10,11,15,16], [12,13,17,18], [14,19], [15,20],[16,17,21,22], [18,19,23,24]],   # X-type stabilizers
    "z": [[0,1], [1,2,6,7], [2,3],[3,4,8,9], [5,6,10,11], [7,8,12,13],  [10,15], [11,12,16,17], [13,14,18,19], [15,16,20,21], [17,18,22,23], [21,22],[23,24]],   # Z-type stabilizers
   
}


# ======================================================
# STEP 2: Measure one stabilizer with the single ancilla
# ======================================================
def measure_x_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    qc.h(anc)
    for q in support:
        qc.cx(anc, data[q])          # ancilla controls, data is target
    qc.h(anc)
    qc.measure(anc, clbit)


def measure_z_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    for q in support:
        qc.cx(data[q], anc)          # data controls, ancilla is target
    qc.measure(anc, clbit)
```

```python
# STEP 3: Build the circuit for a given layout
# ======================================================
def build_circuit(layout, error_type=None, error_qubit=None):
    xs, zs = layout["x"], layout["z"]
    data = QuantumRegister(n_data, "q")
    anc = QuantumRegister(1, "anc")
    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]
    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])

    def one_round(r):
        for i, support in enumerate(xs):
            measure_x_check(qc, data, anc[0], support, sx[r][i])
        for i, support in enumerate(zs):
            measure_z_check(qc, data, anc[0], support, sz[r][i])
        qc.barrier()

    one_round(0)                      # reference round

    if error_type is not None:        # error is injected here
        {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type.upper()](data[error_qubit])
        qc.barrier()

    one_round(1)                      # syndrome round
    return qc
```

```python
# ======================================================
# STEP 4: Run and return the final syndrome arrays
# ======================================================
def run(layout, error_type=None, error_qubit=None):
    qc = build_circuit(layout, error_type, error_qubit)
    counts = AerSimulator().run(qc, shots=1).result().get_counts()
    bitstring = next(iter(counts))
    parts = [[int(b) for b in p[::-1]] for p in bitstring.split()[::-1]]
    x0, z0, x1, z1 = parts
    x_syndrome = [a ^ b for a, b in zip(x0, x1)]
    z_syndrome = [a ^ b for a, b in zip(z0, z1)]
    return x_syndrome, z_syndrome


# ======================================================
# STEP 5: Which stabilizers fired
# ======================================================
def fired_stabilizers(layout, x_syn, z_syn):
    fired_x = [(f"X{i}", layout["x"][i]) for i, b in enumerate(x_syn) if b]
    fired_z = [(f"Z{i}", layout["z"][i]) for i, b in enumerate(z_syn) if b]
    return fired_x, fired_z
```

```python
# ======================================================
# STEP 6: Candidate error locations for one layout
# ======================================================
def locate(fired, checks):
    """Qubits common to ALL fired checks and absent from every unfired check."""
    if not fired:
        return set()
    cand = set.intersection(*[set(checks[i]) for i in fired])
    for i, s in enumerate(checks):
        if i not in fired:
            cand -= set(s)
    return cand


def candidate_errors(layout, x_syn, z_syn):
    """Return ({'X': set, 'Y': set, 'Z': set}, detected)."""
    fired_x = [i for i, b in enumerate(x_syn) if b]
    fired_z = [i for i, b in enumerate(z_syn) if b]
    z_cand = locate(fired_x, layout["x"])    # X-checks fire for Z errors
    x_cand = locate(fired_z, layout["z"])    # Z-checks fire for X errors
    detected = bool(fired_x or fired_z)

    if fired_x and fired_z:
        both = x_cand & z_cand               # qubit common to both groups
        if both:
            return {"Y": both}, detected
        return {k: v for k, v in (("X", x_cand), ("Z", z_cand)) if v}, detected
    if fired_z:
        return ({"X": x_cand} if x_cand else {}), detected
    if fired_x:
        return ({"Z": z_cand} if z_cand else {}), detected
    return {}, detected


def describe(layout, cands, detected):
    if not detected:
        return ["No error detected"]
    if not cands:
        return ["Syndrome does not match a single-qubit error"]
    lines = []
    for kind, cand in cands.items():
        q = sorted(cand)
        text = f"{kind} error at data qubit q" + " or q".join(map(str, q))
        if len(q) > 1:
            stabs = [set(s) for s in layout["x" if kind == "X" else "z"]]
            if kind != "Y" and all({a, b} in stabs for a in q for b in q if a < b):
                text += " (equivalent: they differ by a stabilizer)"
            else:
                text += " (ambiguous)"
        lines.append(text)
    return lines


def is_precise(cands):
    return bool(cands) and all(len(v) == 1 for v in cands.values())
```

```python
# ======================================================
# STEP 7: Decode, and refine with Layout B if ambiguous
# ======================================================
def report(name, layout, x_syn, z_syn):
    print(f"\n--- {name} ---")
    print("X-stabilizer array (Z-error):", x_syn)
    print("Z-stabilizer array (X-error):", z_syn)
    fx, fz = fired_stabilizers(layout, x_syn, z_syn)
    print("X-checks fired:", [f"{n} -> {s}" for n, s in fx] or "none")
    print("Z-checks fired:", [f"{n} -> {s}" for n, s in fz] or "none")
    cands, detected = candidate_errors(layout, x_syn, z_syn)
    for line in describe(layout, cands, detected):
        print("Result:", line)
    return cands, detected


def decode_with_refinement(error_type=None, error_qubit=None):
    xa, za = run(LAYOUT_A, error_type, error_qubit)
    cands_a, det_a = report("Layout A", LAYOUT_A, xa, za)

    if not det_a or is_precise(cands_a):
        print("\nPrecise answer from Layout A, no refinement needed.")
        return cands_a

    print("\nAmbiguous -> re-running with Layout B (new circuit, same error)")
    xb, zb = run(LAYOUT_B, error_type, error_qubit)
    cands_b, _ = report("Layout B", LAYOUT_B, xb, zb)

    final = {k: cands_a[k] & cands_b[k] for k in cands_a if k in cands_b}
    final = {k: v for k, v in final.items() if v}

    print("\n=== Final (Layout A and Layout B combined) ===")
    if not final:
        print("The two layouts disagree: not a single-qubit error.")
    for kind, cand in final.items():
        q = sorted(cand)
        print(f"{kind} error at data qubit q" + " or q".join(map(str, q)))
    return final


if __name__ == "__main__":
    error_type, error_qubit = "z", 8

    qc = build_circuit(LAYOUT_A, error_type, error_qubit)
    print(qc.draw(output="text", fold=150))
    qc.draw(output="mpl", fold=60, filename="surface_code_single_ancilla.png")

    print(f"Injected error: {error_type} on q{error_qubit}")
    decode_with_refinement(error_type, error_qubit)
```

```python
import itertools

LAYOUTS = {"A": LAYOUT_A, "B": LAYOUT_B}


# ======================================================
# STEP 8: Circuit with several injected errors
# ======================================================
def build_circuit_multi(layout, errors=None):
    """errors = list of (type, qubit), e.g. [("X", 2), ("Z", 12)]"""
    xs, zs = layout["x"], layout["z"]
    data = QuantumRegister(n_data, "q")
    anc = QuantumRegister(1, "anc")
    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]
    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])

    def one_round(r):
        for i, support in enumerate(xs):
            measure_x_check(qc, data, anc[0], support, sx[r][i])
        for i, support in enumerate(zs):
            measure_z_check(qc, data, anc[0], support, sz[r][i])
        qc.barrier()

    one_round(0)                                    # reference round

    for error_type, error_qubit in (errors or []):   # all errors injected here
        {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type.upper()](data[error_qubit])
    qc.barrier()

    one_round(1)                                    # syndrome round
    return qc


def run_multi(layout, errors=None):
    qc = build_circuit_multi(layout, errors)
    sim = AerSimulator(method="stabilizer")
    counts = sim.run(qc, shots=1).result().get_counts()
    bitstring = next(iter(counts))
    parts = [[int(b) for b in p[::-1]] for p in bitstring.split()[::-1]]
    x0, z0, x1, z1 = parts
    return [a ^ b for a, b in zip(x0, x1)], [a ^ b for a, b in zip(z0, z1)]


# ======================================================
# STEP 9: Decode by matching against all small error patterns
# ======================================================
def errors_to_pattern(errors):
    """Merge gates on the same qubit (X and Z on one qubit = Y)."""
    bits = {}
    for e, q in errors:
        x, z = {"X": (1, 0), "Z": (0, 1), "Y": (1, 1)}[e.upper()]
        bx, bz = bits.get(q, (0, 0))
        bits[q] = (bx ^ x, bz ^ z)
    letter = {(1, 0): "X", (0, 1): "Z", (1, 1): "Y"}
    return {q: letter[b] for q, b in sorted(bits.items()) if b != (0, 0)}


def pauli_syndrome(layout, pattern):
    """Syndrome a pattern {qubit: 'X'/'Y'/'Z'} would produce."""
    x_syn = [0] * len(layout["x"])
    z_syn = [0] * len(layout["z"])
    for q, e in pattern.items():
        for i, s in enumerate(layout["x"]):
            if q in s and e in "ZY":
                x_syn[i] ^= 1
        for i, s in enumerate(layout["z"]):
            if q in s and e in "XY":
                z_syn[i] ^= 1
    return x_syn, z_syn


def all_patterns(max_weight):
    for w in range(max_weight + 1):
        for qs in itertools.combinations(range(n_data), w):
            for es in itertools.product("XYZ", repeat=w):
                yield dict(zip(qs, es))


# ---- equivalence: two patterns are equivalent if they differ by stabilizers ----
def _mask(qubits):
    m = 0
    for q in qubits:
        m |= 1 << q
    return m


def _basis(vectors):
    basis = []
    for v in vectors:
        for b in basis:
            v = min(v, v ^ b)
        if v:
            basis.append(v)
    return basis


def _in_span(v, basis):
    for b in basis:
        v = min(v, v ^ b)
    return v == 0


_BASES = {}


def equivalent(name, p1, p2):
    if name not in _BASES:
        L = LAYOUTS[name]
        _BASES[name] = (_basis([_mask(s) for s in L["x"]]),
                        _basis([_mask(s) for s in L["z"]]))
    bx, bz = _BASES[name]

    def masks(p):
        xm = _mask(q for q, e in p.items() if e in "XY")
        zm = _mask(q for q, e in p.items() if e in "ZY")
        return xm, zm

    x1, z1 = masks(p1)
    x2, z2 = masks(p2)
    return _in_span(x1 ^ x2, bx) and _in_span(z1 ^ z2, bz)


def group_equivalent(name, patterns):
    classes = []
    for p in patterns:
        for c in classes:
            if equivalent(name, c[0], p):
                c.append(p)
                break
        else:
            classes.append([p])
    return classes


def decode_multi(measured, max_weight=2):
    """
    measured = {"A": (x_syn, z_syn)}  and optionally "B": (...)
    Returns classes of lowest-weight patterns that explain ALL measured syndromes.
    """
    matches = [p for p in all_patterns(max_weight)
               if all(pauli_syndrome(LAYOUTS[n], p) == syn
                      for n, syn in measured.items())]
    if not matches:
        return []
    w = min(len(p) for p in matches)
    best = [p for p in matches if len(p) == w]
    return group_equivalent(next(iter(measured)), best)


def pattern_text(p):
    return ", ".join(f"{e} on q{q}" for q, e in p.items()) or "no error"


def show_syndrome(name, layout, x_syn, z_syn):
    print(f"\n--- Layout {name} ---")
    print("X-stabilizer array (Z-error):", x_syn)
    print("Z-stabilizer array (X-error):", z_syn)
    fx, fz = fired_stabilizers(layout, x_syn, z_syn)
    print("X-checks fired:", [f"{n} -> {s}" for n, s in fx] or "none")
    print("Z-checks fired:", [f"{n} -> {s}" for n, s in fz] or "none")


def detect_multi(errors, max_weight=2, layouts=("A",)):
    """layouts=("A",) is enough for 2 errors; use ("A", "B") to pinpoint exactly."""
    print("Injected :", pattern_text(errors_to_pattern(errors)))

    measured = {}
    for name in layouts:
        x_syn, z_syn = run_multi(LAYOUTS[name], errors)
        measured[name] = (x_syn, z_syn)
        show_syndrome(name, LAYOUTS[name], x_syn, z_syn)

    classes = decode_multi(measured, max_weight)

    print("\n=== Decoded (lowest-weight match) ===")
    if not classes:
        print(f"No pattern with up to {max_weight} errors matches the syndrome.")
    for c in classes:
        print(" ", pattern_text(c[0]))
        if len(c) > 1:
            others = "; ".join(pattern_text(p) for p in c[1:4])
            more = " ..." if len(c) > 4 else ""
            print(f"    ({len(c) - 1} equivalent pattern(s), same correction works: "
                  f"{others}{more})")
    if len(classes) > 1:
        print(f"Ambiguous: {len(classes)} genuinely different possibilities.")
    return classes
```

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister

n_data = 25
# 5x5 grid, qubit index = 5*row + col
#   q0  q1  q2  q3  q4
#   q5  q6  q7  q8  q9
#   q10 q11 q12 q13 q14
#   q15 q16 q17 q18 q19
#   q20 q21 q22 q23 q24

LAYOUT_A = {
    "x": [[0,1,5,6],[1,2],[2,3,7,8],[3,4],[6,7,11,12],[8,9,13,14],
          [10,11,15,16],[12,13,17,18],[16,17,21,22],[18,19,23,24],[20,21],[22,23]],
    "z": [[0,5],[1,2,6,7],[3,4,8,9],[5,6,10,11],[7,8,12,13],[9,14],
          [10,15],[11,12,16,17],[13,14,18,19],[15,16,20,21],[17,18,22,23],[19,24]],
}
LAYOUT_B = {
    "x": [[0,1,5,6],[2,3,7,8],[4,9],[5,10],[6,7,11,12],[8,9,13,14],
          [10,11,15,16],[12,13,17,18],[14,19],[15,20],[16,17,21,22],[18,19,23,24]],
    "z": [[0,1],[1,2,6,7],[2,3],[3,4,8,9],[5,6,10,11],[7,8,12,13],
          [11,12,16,17],[13,14,18,19],[15,16,20,21],[17,18,22,23],[21,22],[23,24]],
}

# Sanity check: every X and Z stabilizer must overlap on an even number of qubits
def check_commutation(layout):
    bad = [(x, z) for x in layout["x"] for z in layout["z"]
           if len(set(x) & set(z)) % 2]
    print("OK" if not bad else f"Anticommuting pairs: {bad}")

check_commutation(LAYOUT_A)
check_commutation(LAYOUT_B)
```

```python
# Measure one stabilizer with the single ancilla (reset first, then measure)
def measure_x_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    qc.h(anc)
    for q in support:
        qc.cx(anc, data[q])      # ancilla controls, data is target
    qc.h(anc)
    qc.measure(anc, clbit)

def measure_z_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    for q in support:
        qc.cx(data[q], anc)      # data controls, ancilla is target
    qc.measure(anc, clbit)
```

```python
# Build one round of syndrome extraction for a given layout
def build_round_circuit(layout):
    xs, zs = layout["x"], layout["z"]
    data = QuantumRegister(n_data, "q")
    anc  = QuantumRegister(1, "anc")
    cx_  = ClassicalRegister(len(xs), "x_syn")
    cz_  = ClassicalRegister(len(zs), "z_syn")
    qc = QuantumCircuit(data, anc, cx_, cz_)
    for i, s in enumerate(xs):
        measure_x_check(qc, data, anc[0], s, cx_[i])
    for i, s in enumerate(zs):
        measure_z_check(qc, data, anc[0], s, cz_[i])
    return qc

qc_A = build_round_circuit(LAYOUT_A)
qc_B = build_round_circuit(LAYOUT_B)
print(qc_A.num_qubits, qc_A.num_clbits, "|", qc_B.num_qubits, qc_B.num_clbits)
```

```python
from qiskit_aer import AerSimulator

def build_circuit(layout, error_type=None, error_qubit=None):
    xs, zs = layout["x"], layout["z"]
    data = QuantumRegister(n_data, "q")
    anc  = QuantumRegister(1, "anc")
    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]
    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])

    def one_round(r):
        for i, s in enumerate(xs):
            measure_x_check(qc, data, anc[0], s, sx[r][i])
        for i, s in enumerate(zs):
            measure_z_check(qc, data, anc[0], s, sz[r][i])
        qc.barrier()

    one_round(0)                                   # round 1 (reference)
    if error_type is not None:                     # error injected here
        {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type.upper()](data[error_qubit])
        qc.barrier()
    one_round(1)                                   # round 2
    return qc

def run(layout, error_type=None, error_qubit=None):
    qc = build_circuit(layout, error_type, error_qubit)
    sim = AerSimulator(method="stabilizer")        # all-Clifford, so this is fast
    counts = sim.run(qc, shots=1).result().get_counts()
    bitstring = next(iter(counts))
    # registers print in reverse order: z_round1 x_round1 z_round0 x_round0
    x0, z0, x1, z1 = [[int(b) for b in p[::-1]] for p in bitstring.split()[::-1]]
    x_syn = [a ^ b for a, b in zip(x0, x1)]        # XOR of the two rounds
    z_syn = [a ^ b for a, b in zip(z0, z1)]
    return x_syn, z_syn

def fired_stabilizers(layout, x_syn, z_syn):
    fx = [(f"X{i}", layout["x"][i]) for i, b in enumerate(x_syn) if b]
    fz = [(f"Z{i}", layout["z"][i]) for i, b in enumerate(z_syn) if b]
    return fx, fz
def candidates_from_one_array(stabs, syn):
    """stabs = list of supports for one type (X or Z), syn = its syndrome array."""
    fired = [set(s) for s, b in zip(stabs, syn) if b]
    if not fired:
        return None                                  # this type did not fire

    # qubits common to every fired stabilizer
    cand = set.intersection(*fired)

    # remove any qubit that also sits in a stabilizer that did NOT fire
    for s, b in zip(stabs, syn):
        if not b:
            cand -= set(s)
    return cand

def locate_error(layout, x_syn, z_syn):
    cx = candidates_from_one_array(layout["x"], x_syn)
    cz = candidates_from_one_array(layout["z"], z_syn)

    if cx is None and cz is None:
        return "No error detected"
    if cx is not None and cz is not None:            # two arrays fired (Y error)
        return sorted(cx & cz)                       # common element of both
    return sorted(cx if cx is not None else cz)      # single array fired
# ---- example: X error on qubit 6 in layout A ----

x_syn, z_syn = run(LAYOUT_A, "z", 16)
fx, fz = fired_stabilizers(LAYOUT_A, x_syn, z_syn)
print("X syndrome array:", x_syn)
print("Z syndrome array:", z_syn)
print("Fired X stabilizers:", fx)
print("Fired Z stabilizers:", fz)
```

```python
def classify_and_locate(layout, x_syn, z_syn):
    """
    Returns (case, candidates) for ONE layout.
      case 1: exactly one stabilizer fired  -> usually several candidates (ambiguous)
      case 2: two or more stabilizers fired -> intersection usually gives one qubit
    Candidates combine X-type and Z-type results (a Y error fires both).
    """
    n_fired = sum(x_syn) + sum(z_syn)
    if n_fired == 0:
        return "no error", set()

    cx = candidates_from_one_array(layout["x"], x_syn)
    cz = candidates_from_one_array(layout["z"], z_syn)

    if cx is not None and cz is not None:       # Y error: both arrays fired
        cand = cx & cz
    else:
        cand = cx if cx is not None else cz

    # fired count within the type that fired (for the case label)
    case = "case 1 (single stabilizer)" if max(sum(x_syn), sum(z_syn)) == 1 \
           else "case 2 (multiple stabilizers)"
    return case, cand


def locate_error_two_layouts(error_type, error_qubit, verbose=True):
    """Run the same error through layout A and layout B, then combine."""
    xa, za = run(LAYOUT_A, error_type, error_qubit)
    xb, zb = run(LAYOUT_B, error_type, error_qubit)

    case_a, cand_a = classify_and_locate(LAYOUT_A, xa, za)
    case_b, cand_b = classify_and_locate(LAYOUT_B, xb, zb)

    if verbose:
        print(f"Layout A: {case_a}, candidates = {sorted(cand_a)}")
        print(f"Layout B: {case_b}, candidates = {sorted(cand_b)}")

    # Case 2 in a layout: already a single qubit -> done.
    if len(cand_a) == 1:
        final = cand_a
    elif len(cand_b) == 1:
        final = cand_b
    # Case 1 in both layouts: intersect the two ambiguous sets.
    else:
        final = cand_a & cand_b

    return sorted(final)


# ---- Case 1: single stabilizer fires in A -> need layout B ----
# Z error on qubit 0: A fires only [0,1,5,6] -> candidates {0,5}
#                     B fires only [0,1,5,6] -> candidates {0,1}
#                     intersection -> {0}
print("Z error on q0 ->", locate_error_two_layouts("x", 1))

# ---- Case 2: two stabilizers fire in A -> intersection is enough ----
# Z error on qubit 2: A fires [1,2] and [2,3,7,8] -> common qubit {2}
print("Z error on q2 ->", locate_error_two_layouts("Z", 1))

# ---- Full verification over every qubit and every error type ----
failures = []
for q in range(n_data):
    for e in "XYZ":
        if locate_error_two_layouts(e, q, verbose=False) != [q]:
            failures.append((e, q))
print("Failures:", failures if failures else "none, all 75 single-qubit errors located")
```

```python
def candidates_for_type(stabs, syn):
    """Candidate error qubits from ONE syndrome array in ONE layout.
    Case 1: one stabilizer fired  -> its not_common (unique) qubits
    Case 2: 2+ stabilizers fired  -> qubits common to all fired stabilizers
    """
    fired = [i for i, b in enumerate(syn) if b]
    if not fired:
        return None, "not fired"

    if len(fired) == 1:                                   # case 1
        i = fired[0]
        others = {q for j, s in enumerate(stabs) if j != i for q in s}
        return {q for q in stabs[i] if q not in others}, f"case 1 (only stabilizer {i} fired)"

    common = set.intersection(*[set(stabs[i]) for i in fired])   # case 2
    return common, f"case 2 (stabilizers {fired} fired)"


def locate_final(error_type, error_qubit):
    xa, za = run(LAYOUT_A, error_type, error_qubit)
    xb, zb = run(LAYOUT_B, error_type, error_qubit)
    print(f"\n{error_type} error on q{error_qubit}")

    final = None
    # X-type checks detect Z errors, Z-type checks detect X errors
    for label, stabs_key, syn_A, syn_B in (("X-checks (Z part)", "x", xa, xb),
                                           ("Z-checks (X part)", "z", za, zb)):
        cA, ruleA = candidates_for_type(LAYOUT_A[stabs_key], syn_A)
        cB, ruleB = candidates_for_type(LAYOUT_B[stabs_key], syn_B)
        if cA is None and cB is None:
            continue
        both = cA & cB                                    # intersect the two layouts
        print(f"  {label}")
        print(f"    layout A: {sorted(cA)}  <- {ruleA}")
        print(f"    layout B: {sorted(cB)}  <- {ruleB}")
        print(f"    common  : {sorted(both)}")
        final = both if final is None else final & both   # Y error: combine both types

    print("  final error qubit:", sorted(final) if final is not None else "No error detected")
    return final


# ---- tests ----
locate_final("Z", 0)    # case 1: only [0,1,5,6] fires in both layouts
locate_final("Z", 2)    # case 2 in A ([1,2] and [2,3,7,8] fire)
locate_final("Z", 16)   # case 2 in both layouts
locate_final("X", 6)    # Z-checks side
locate_final("Y", 0)    # both arrays fire
```

```python
def common_with_others(layout, x_syn, z_syn):
    """For each fired stabilizer, list its qubits that also appear in other
    stabilizers of the same type (X with X, Z with Z)."""
    result = {}
    for label, stabs, syn in (("X", layout["x"], x_syn), ("Z", layout["z"], z_syn)):
        for i, b in enumerate(syn):
            if not b:
                continue
            shared = {}
            for q in stabs[i]:
                others = [j for j, s in enumerate(stabs) if j != i and q in s]
                if others:
                    shared[q] = [f"{label}{j}" for j in others]
            result[f"{label}{i}"] = {
                "support": stabs[i],
                "common_qubits": shared,                       # qubit -> other stabilizers containing it
                "not_common": [q for q in stabs[i] if q not in shared],  # unique qubits
            }
    return result

# ---- example ----
x_syn, z_syn = run(LAYOUT_A, "Z", 16)
fx, fz = fired_stabilizers(LAYOUT_A, x_syn, z_syn)
print("X syndrome array:", x_syn)
print("Z syndrome array:", z_syn)
print("Fired X stabilizers:", fx)
print("Fired Z stabilizers:", fz)
for name, info in common_with_others(LAYOUT_A, x_syn, z_syn).items():
    print(name, info)

# Same check as before, now on LAYOUT_B (Z error on qubit 0)
x_syn, z_syn = run(LAYOUT_B, "Z", 16)
fx, fz = fired_stabilizers(LAYOUT_B, x_syn, z_syn)

print("X syndrome array:", x_syn)
print("Z syndrome array:", z_syn)
print("Fired X stabilizers:", fx)
print("Fired Z stabilizers:", fz)
for name, info in common_with_others(LAYOUT_B, x_syn, z_syn).items():
    print(name, info)    
```

```python

from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator


# ============================================================
# 1. Build two measurement rounds, with optional injected error
# ============================================================

def build_circuit(layout, error_type=None, error_qubit=None):

    xs, zs = layout["x"], layout["z"]

    data = QuantumRegister(n_data, "q")
    anc = QuantumRegister(1, "anc")

    # Separate classical registers for each round
    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]

    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])

    def one_round(r):
        # Measure all X stabilizers
        for i, support in enumerate(xs):
            measure_x_check(
                qc, data, anc[0], support, sx[r][i]
            )

        # Measure all Z stabilizers
        for i, support in enumerate(zs):
            measure_z_check(
                qc, data, anc[0], support, sz[r][i]
            )

        qc.barrier()

    # Round 0: reference syndrome
    one_round(0)

    # Inject an error between the two rounds
    if error_type is not None:

        error_type = error_type.upper()

        if error_type not in ("X", "Y", "Z"):
            raise ValueError("error_type must be X, Y, Z or None")

        if error_qubit is None or not 0 <= error_qubit < n_data:
            raise ValueError("Provide a valid data-qubit index")

        {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type](
            data[error_qubit]
        )

        qc.barrier()

    # Round 1: syndrome after the possible error
    one_round(1)

    return qc


# ============================================================
# 2. Run the circuit and calculate the syndrome difference
# ============================================================

def run(layout, error_type=None, error_qubit=None):

    qc = build_circuit(layout, error_type, error_qubit)

    sim = AerSimulator(method="stabilizer")

    counts = sim.run(qc, shots=1).result().get_counts(qc)
    bitstring = next(iter(counts))

    # Classical registers appear in reverse order in count strings:
    # z_round1 x_round1 z_round0 x_round0
    parts = bitstring.split()

    if len(parts) != 4:
        raise RuntimeError(
            f"Expected 4 classical-register groups, got: {bitstring}"
        )

    z1, x1, z0, x0 = [
        [int(b) for b in part[::-1]]
        for part in parts
    ]

    # Compare the two rounds bit by bit
    x_syn = [a ^ b for a, b in zip(x0, x1)]
    z_syn = [a ^ b for a, b in zip(z0, z1)]

    return x_syn, z_syn


# ============================================================
# 3. Identify which stabilizers changed
# ============================================================

def fired_stabilizers(layout, x_syn, z_syn):

    fx = [
        (f"X{i}", layout["x"][i])
        for i, bit in enumerate(x_syn) if bit
    ]

    fz = [
        (f"Z{i}", layout["z"][i])
        for i, bit in enumerate(z_syn) if bit
    ]

    return fx, fz


# ============================================================
# 4. Find candidate qubits from one syndrome array
# ============================================================

def candidates_for_type(stabs, syn):

    # Supports of stabilizers whose syndrome changed
    fired = [
        set(support)
        for support, bit in zip(stabs, syn)
        if bit
    ]

    if not fired:
        return None

    # A candidate must belong to every fired stabilizer
    candidates = set.intersection(*fired)

    # It must not belong to any stabilizer that did not fire
    for support, bit in zip(stabs, syn):
        if not bit:
            candidates -= set(support)

    return candidates


# ============================================================
# 5. Compare layouts A and B, and combine X/Z information
# ============================================================

def locate_final(error_type=None, error_qubit=None):

    # Run the same error experiment on both layouts
    xa, za = run(LAYOUT_A, error_type, error_qubit)
    xb, zb = run(LAYOUT_B, error_type, error_qubit)

    print(f"\nInjected error: {error_type} on q{error_qubit}")

    final = None

    checks_to_process = [
        ("X-checks (detect Z component)", "x", xa, xb),
        ("Z-checks (detect X component)", "z", za, zb),
    ]

    for label, key, syn_a, syn_b in checks_to_process:

        cA = candidates_for_type(LAYOUT_A[key], syn_a)
        cB = candidates_for_type(LAYOUT_B[key], syn_b)

        # Neither layout detected a change in this check type
        if cA is None and cB is None:
            continue

        print(f"\n{label}")
        print("  Layout A candidates:", sorted(cA) if cA is not None else None)
        print("  Layout B candidates:", sorted(cB) if cB is not None else None)

        # Need evidence from both layouts to localize by intersection
        if cA is None or cB is None:
            print("  Common candidates: unavailable")
            continue

        common = cA & cB

        print("  Common candidates:", sorted(common))

        if final is None:
            final = common
        else:
            # For a Y error, both check types must point to the
            # same data qubit
            final &= common

    result = sorted(final) if final else (
        "No error detected" if final is None else []
    )

    print("\nFinal candidate:", result)

    return result


# ============================================================
# 6. Test cases
# ============================================================

locate_final("Z", 0)
locate_final("Z", 2)
locate_final("Z", 16)
locate_final("X", 6)
locate_final("Y", 0)
locate_final()             # No injected error
```

```python

"""
5x5 stabilizer code: single-ancilla syndrome extraction + two-layout error locator.
 
Qubit index = 5*row + col
 
     q0  q1  q2  q3  q4
     q5  q6  q7  q8  q9
     q10 q11 q12 q13 q14
     q15 q16 q17 q18 q19
     q20 q21 q22 q23 q24
 
Rules used by the decoder:
  * X-type stabilizers detect Z errors.
  * Z-type stabilizers detect X errors.
  * A Y error (= X and Z on the same qubit) fires both types.
"""
 
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
 
# ======================================================================
# 1. Layouts
# ======================================================================
N_DATA = 25
 
LAYOUT_A = {
    "x": [[0, 1, 5, 6], [1, 2], [2, 3, 7, 8], [3, 4], [6, 7, 11, 12], [8, 9, 13, 14],
          [10, 11, 15, 16], [12, 13, 17, 18], [16, 17, 21, 22], [18, 19, 23, 24],
          [20, 21], [22, 23]],
    "z": [[0, 5], [1, 2, 6, 7], [3, 4, 8, 9], [5, 6, 10, 11], [7, 8, 12, 13], [9, 14],
          [10, 15], [11, 12, 16, 17], [13, 14, 18, 19], [15, 16, 20, 21],
          [17, 18, 22, 23], [19, 24]],
}
 
LAYOUT_B = {
    "x": [[0, 1, 5, 6], [2, 3, 7, 8], [4, 9], [5, 10], [6, 7, 11, 12], [8, 9, 13, 14],
          [10, 11, 15, 16], [12, 13, 17, 18], [14, 19], [15, 20], [16, 17, 21, 22],
          [18, 19, 23, 24]],
    "z": [[0, 1], [1, 2, 6, 7], [2, 3], [3, 4, 8, 9], [5, 6, 10, 11], [7, 8, 12, 13],
          [11, 12, 16, 17], [13, 14, 18, 19], [15, 16, 20, 21], [17, 18, 22, 23],
          [21, 22], [23, 24]],
}
 
 
def check_commutation(layout, name=""):
    """Every X and Z stabilizer must overlap on an even number of qubits."""
    bad = [(x, z) for x in layout["x"] for z in layout["z"]
           if len(set(x) & set(z)) % 2]
    print(f"Layout {name}: " + ("commutation OK" if not bad else f"anticommuting pairs: {bad}"))
 
 
# ======================================================================
# 2. Circuit construction (one reusable ancilla)
# ======================================================================
def measure_x_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    qc.h(anc)
    for q in support:
        qc.cx(anc, data[q])          # ancilla controls, data is target
    qc.h(anc)
    qc.measure(anc, clbit)
 
 
def measure_z_check(qc, data, anc, support, clbit):
    qc.reset(anc)
    for q in support:
        qc.cx(data[q], anc)          # data controls, ancilla is target
    qc.measure(anc, clbit)
 
 
def build_circuit(layout, error_type=None, error_qubit=None):
    """Two syndrome rounds with an optional Pauli error injected in between."""
    xs, zs = layout["x"], layout["z"]
    data = QuantumRegister(N_DATA, "q")
    anc = QuantumRegister(1, "anc")
    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]
    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])
 
    def one_round(r):
        for i, s in enumerate(xs):
            measure_x_check(qc, data, anc[0], s, sx[r][i])
        for i, s in enumerate(zs):
            measure_z_check(qc, data, anc[0], s, sz[r][i])
        qc.barrier()
 
    one_round(0)                                     # reference round
    if error_type is not None:                       # error injected here
        {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type.upper()](data[error_qubit])
        qc.barrier()
    one_round(1)                                     # detection round
    return qc
 
 
def run(layout, error_type=None, error_qubit=None):
    """Simulate and return (x_syndrome, z_syndrome) = XOR of the two rounds."""
    qc = build_circuit(layout, error_type, error_qubit)
    sim = AerSimulator(method="stabilizer")          # all-Clifford -> fast
    counts = sim.run(qc, shots=1).result().get_counts()
    bitstring = next(iter(counts))
    # Qiskit prints registers in reverse: z_round1 x_round1 z_round0 x_round0
    x0, z0, x1, z1 = [[int(b) for b in p[::-1]] for p in bitstring.split()[::-1]]
    x_syn = [a ^ b for a, b in zip(x0, x1)]
    z_syn = [a ^ b for a, b in zip(z0, z1)]
    return x_syn, z_syn
```

```python
# ======================================================================
# 3. Diagnostics (optional helpers)
# ======================================================================
def fired_stabilizers(layout, x_syn, z_syn):
    fx = [(f"X{i}", layout["x"][i]) for i, b in enumerate(x_syn) if b]
    fz = [(f"Z{i}", layout["z"][i]) for i, b in enumerate(z_syn) if b]
    return fx, fz
 
 
def common_with_others(layout, x_syn, z_syn):
    """For each fired stabilizer, show which of its qubits are shared with
    other stabilizers of the same type, and which are unique to it."""
    result = {}
    for label, stabs, syn in (("X", layout["x"], x_syn), ("Z", layout["z"], z_syn)):
        for i, b in enumerate(syn):
            if not b:
                continue
            shared = {}
            for q in stabs[i]:
                others = [j for j, s in enumerate(stabs) if j != i and q in s]
                if others:
                    shared[q] = [f"{label}{j}" for j in others]
            result[f"{label}{i}"] = {
                "support": stabs[i],
                "common_qubits": shared,
                "not_common": [q for q in stabs[i] if q not in shared],
            }
    return result
 
```

```python
# 4. Decoder
# ======================================================================
def candidates_from_one_array(stabs, syn):
    """Candidate qubits for one stabilizer type (X or Z) in one layout.
 
    Case 1 (one stabilizer fired): the qubits unique to it (usually 2 -> ambiguous).
    Case 2 (several fired):        the qubit common to all of them.
    Returns None if nothing fired.
    """
    fired = [set(s) for s, b in zip(stabs, syn) if b]
    if not fired:
        return None
    cand = set.intersection(*fired)                  # common to every fired stabilizer
    for s, b in zip(stabs, syn):
        if not b:
            cand -= set(s)                           # drop qubits seen by silent stabilizers
    return cand
 
 
def locate_part(stabs_A, syn_A, stabs_B, syn_B):
    """Locate one error component (X-part or Z-part) using both layouts.
    Returns a qubit index, a set of unresolved candidates, or None if nothing fired."""
    cA = candidates_from_one_array(stabs_A, syn_A)
    cB = candidates_from_one_array(stabs_B, syn_B)
    if cA is None and cB is None:
        return None
    if cA is not None and len(cA) == 1:              # case 2: layout A alone is enough
        return next(iter(cA))
    if cB is not None and len(cB) == 1:              # case 2 in layout B
        return next(iter(cB))
    final = (cA or set()) & (cB or set())            # case 1: intersect both layouts
    return next(iter(final)) if len(final) == 1 else final
 
 
def locate_error_two_layouts(error_type, error_qubit, verbose=True):
    """Inject an error, run both layouts, and report the error type and qubit."""
    xa, za = run(LAYOUT_A, error_type, error_qubit)
    xb, zb = run(LAYOUT_B, error_type, error_qubit)
 
    # X-type stabilizers detect Z errors; Z-type stabilizers detect X errors
    z_part = locate_part(LAYOUT_A["x"], xa, LAYOUT_B["x"], xb)
    x_part = locate_part(LAYOUT_A["z"], za, LAYOUT_B["z"], zb)
 
    if verbose:
        print(f"Layout A  X-syn={xa}\n          Z-syn={za}")
        print(f"Layout B  X-syn={xb}\n          Z-syn={zb}")
 
    if x_part is None and z_part is None:
        return "No error detected"
    if x_part is not None and z_part is not None:
        if x_part == z_part:
            return f"Y error on qubit {x_part}"      # X and Z on the same qubit
        return f"X error on qubit {x_part} and Z error on qubit {z_part}"
    if x_part is not None:
        return f"X error on qubit {x_part}"
    return f"Z error on qubit {z_part}"
 
 
```

```python
# ======================================================================
# 5. Reporting: measured syndromes + stabilizer sets that fired
# ======================================================================
def report(error_type, error_qubit):
    """Print, for both layouts, the syndrome arrays and the stabilizer sets that fired,
    then the decoded result."""
    print("=" * 70)
    print(f"Injected: {error_type.upper()} error on qubit {error_qubit}")
    print("=" * 70)
 
    for name, layout in (("A", LAYOUT_A), ("B", LAYOUT_B)):
        x_syn, z_syn = run(layout, error_type, error_qubit)
        fx, fz = fired_stabilizers(layout, x_syn, z_syn)
 
        print(f"\nLayout {name}")
        print(f"  X syndrome array : {x_syn}")
        print(f"  Z syndrome array : {z_syn}")
 
        print("  Fired X stabilizers (detect Z errors):")
        print("".join(f"    {lbl}: {sup}\n" for lbl, sup in fx) if fx else "    none")
        print("  Fired Z stabilizers (detect X errors):")
        print("".join(f"    {lbl}: {sup}\n" for lbl, sup in fz) if fz else "    none")
 
        cx = candidates_from_one_array(layout["x"], x_syn)
        cz = candidates_from_one_array(layout["z"], z_syn)
        print(f"  Candidates from X-type (Z-error part): {sorted(cx) if cx is not None else None}")
        print(f"  Candidates from Z-type (X-error part): {sorted(cz) if cz is not None else None}")
 
    print("\nDecoded result:", locate_error_two_layouts(error_type, error_qubit, verbose=False))
    print()
 
```

```python
# ======================================================================
# 6. Tests
# ======================================================================

def run(layout, errors=None):
    """
    Simulate two syndrome rounds with multiple injected errors.
    Return (x_syndrome, z_syndrome).
    """

    qc = build_circuit(layout, errors=errors)

    sim = AerSimulator(method="stabilizer")
    counts = sim.run(qc, shots=1).result().get_counts()

    bitstring = next(iter(counts))

    # Qiskit displays classical registers in reverse order
    x0, z0, x1, z1 = [
        [int(b) for b in part[::-1]]
        for part in bitstring.split()[::-1]
    ]

    # Detection events = syndrome in round 0 XOR round 1
    x_syn = [a ^ b for a, b in zip(x0, x1)]
    z_syn = [a ^ b for a, b in zip(z0, z1)]

    return x_syn, z_syn

def verify_all():
    """Verify localization of all 75 single-qubit Pauli errors."""
    failures = []

    for q in range(N_DATA):
        for e in "XYZ":
            result = locate_error_two_layouts(e, q, verbose=False)
            expected = f"{e} error on qubit {q}"

            if result != expected:
                failures.append((e, q, result, expected))

    print(
        "Failures:",
        failures if failures
        else "none — all 75 single-qubit errors located correctly"
    )
```

```python

def print_detected_stabilizers(layout, x_syndrome, z_syndrome):
    """Print stabilizers whose detection-event bit is 1."""

    detected = False

    print("\nDetected stabilizers:")

    # X-type stabilizers detect Z-error components
    for i, bit in enumerate(x_syndrome):
        if bit == 1:
            print(f"  X{i}: support = {layout['x'][i]}")
            detected = True

    # Z-type stabilizers detect X-error components
    for i, bit in enumerate(z_syndrome):
        if bit == 1:
            print(f"  Z{i}: support = {layout['z'][i]}")
            detected = True

    if not detected:
        print("  No syndrome changes detected.")


def test_errors(errors=None, separate=False):
    """
    Test Pauli errors on layouts A and B.

    errors:
        [("X", 2), ("Z", 7)]

    separate=False: inject all errors simultaneously.
    separate=True: test each error in its own circuit.
    """

    if errors is None:
        errors = []

    # Validate all errors
    normalized_errors = []

    for error_type, qubit in errors:
        error_type = error_type.upper()

        if error_type not in ("X", "Y", "Z"):
            raise ValueError(
                f"Invalid error type {error_type}. Use X, Y, or Z."
            )

        if not isinstance(qubit, int) or not 0 <= qubit < N_DATA:
            raise ValueError(
                f"Qubit must be an integer from 0 to {N_DATA - 1}."
            )

        normalized_errors.append((error_type, qubit))

    errors = normalized_errors

    print("\nInjected errors:", errors if errors else "None")
    print("Mode:", "separate circuits" if separate else "simultaneous errors")

    for name, layout in (("A", LAYOUT_A), ("B", LAYOUT_B)):
        print(f"\n{'=' * 50}")
        print(f"Layout {name}")
        print(f"{'=' * 50}")

        if not errors:
            test_cases = [("No injected error", [])]
        elif separate:
            test_cases = [
                (f"{e} error on q{q}", [(e, q)])
                for e, q in errors
            ]
        else:
            description = ", ".join(
                f"{e} on q{q}" for e, q in errors
            )
            test_cases = [(description, errors)]

        for description, case_errors in test_cases:
            x_syn, z_syn = run(layout, errors=case_errors)

            print(f"\nTest: {description}")
            print("X syndrome:", x_syn)
            print("Z syndrome:", z_syn)

            print_detected_stabilizers(
                layout,
                x_syn,
                z_syn
            )


# ==========================================================
# TEST 1: Multiple errors simultaneously
# ==========================================================
test_errors([
    ("X", 5),
    ("Z", 10),
])


# ==========================================================
# TEST 2: Test those errors separately
# Uncomment to run this test too.
# ==========================================================
# test_errors([
#     ("X", 2),
#     ("Z", 7),
# ], separate=True)


# ==========================================================
# TEST 3: Y error
# ==========================================================
# test_errors([("Y", 6)])


# ==========================================================
# TEST 4: No injected error
# ==========================================================
# test_errors([])
```

```python

def build_circuit(layout, errors=None):
    """
    Two syndrome rounds with multiple Pauli errors
    injected between the rounds.
    """
    xs, zs = layout["x"], layout["z"]

    data = QuantumRegister(N_DATA, "q")
    anc = QuantumRegister(1, "anc")

    sx = [ClassicalRegister(len(xs), f"x_round{r}") for r in range(2)]
    sz = [ClassicalRegister(len(zs), f"z_round{r}") for r in range(2)]

    qc = QuantumCircuit(data, anc, sx[0], sz[0], sx[1], sz[1])

    if errors is None:
        errors = [("X", 23), ("Z", 10)]

    # Validate errors
    for error_type, error_qubit in errors:
        error_type = error_type.upper()

        if error_type not in ("X", "Y", "Z"):
            raise ValueError(f"Invalid error type: {error_type}")

        if not isinstance(error_qubit, int) or not 0 <= error_qubit < N_DATA:
            raise ValueError(f"Invalid data qubit: {error_qubit}")

    def one_round(r):
        for i, support in enumerate(xs):
            measure_x_check(qc, data, anc[0], support, sx[r][i])

        for i, support in enumerate(zs):
            measure_z_check(qc, data, anc[0], support, sz[r][i])

        qc.barrier()

    # Round 0: reference syndrome
    one_round(0)

    # Inject multiple errors
    if errors:
        qc.barrier(label="Before errors")

        for error_type, error_qubit in errors:
            gate = {"X": qc.x, "Y": qc.y, "Z": qc.z}[error_type.upper()]
            gate(data[error_qubit])
            qc.barrier(label=f"{error_type.upper()} on q{error_qubit}")

    # Round 1: syndrome after errors
    one_round(1)

    return qc


# ==========================================================
# PRINT THE QUANTUM CIRCUIT — outside build_circuit()
# ==========================================================

# errors = [("X", 23), ("Z", 10)]

# qc = build_circuit(LAYOUT_A, errors=errors)

# print("\nQuantum Circuit for Layout A:")
# print(qc.draw(output="text", fold=150))


# # Optional: graphical circuit in Jupyter Notebook
# from IPython.display import display
# display(qc.draw(output="mpl", fold=60))


# # ==========================================================
# # SIMULATE BOTH LAYOUTS
# # ==========================================================

# for name, layout in (("A", LAYOUT_A), ("B", LAYOUT_B)):
#     x_syn, z_syn = run(layout, errors=errors)

#     print(f"\nLayout {name}")
#     print("X syndrome:", x_syn)
#     print("Z syndrome:", z_syn)
errors = [("X", 5), ("Z", 10)]

qc = build_circuit(LAYOUT_A, errors=errors)

print("\nQuantum Circuit for Layout A:")
print(qc.draw(output="text", fold=150))

from IPython.display import display
display(qc.draw(output="mpl", fold=60))

for name, layout in (("A", LAYOUT_A), ("B", LAYOUT_B)):
    x_syn, z_syn = run(layout, errors=errors)

    print(f"\nLayout {name}")
    print("X syndrome:", x_syn)
    print("Z syndrome:", z_syn)
```
